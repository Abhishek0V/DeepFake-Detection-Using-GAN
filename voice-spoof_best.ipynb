## preprocessing ----

```python
TARGET_SR = 16000
FIXED_DURATION = 4  # seconds
FIXED_LENGTH = TARGET_SR * FIXED_DURATION

def preprocess_audio(file_path):
    # Load audio (auto converts to mono)
    audio, sr = librosa.load(file_path, sr=TARGET_SR, mono=True)

    # Remove NaN
    audio = np.nan_to_num(audio)

    # Peak normalization
    peak = np.max(np.abs(audio))
    if peak > 0:
        audio = audio / peak

    # Fix length (pad or trim)
    # Add this line BEFORE the fix length block
    audio, _ = librosa.effects.trim(audio, top_db=30)

    # Then your existing fix length code
    if len(audio) > FIXED_LENGTH:
        audio = audio[:FIXED_LENGTH]
    else:
        pad_width = FIXED_LENGTH - len(audio)
        audio = np.pad(audio, (0, pad_width))

    return audio
```

# Feature Extraction

```python
from spafe.features.lfcc import lfcc
import librosa
import numpy as np

def extract_lfcc(file_path):
    # Load directly from flac — no intermediate .wav needed
    audio, sr = librosa.load(file_path, sr=TARGET_SR, mono=True)

    # Preprocess
    audio = preprocess_audio(file_path)  # your existing function

    # LFCC
    feats = lfcc(
        sig=audio,
        fs=16000,
        num_ceps=60,
        nfilts=70,
        nfft=512,
        low_freq=0,
        high_freq=8000
    ).astype(np.float32)

    # ✅ Delta BEFORE CMVN
    delta  = librosa.feature.delta(feats)
    delta2 = librosa.feature.delta(feats, order=2)

    # Concatenate all
    feats = np.concatenate([feats, delta, delta2], axis=1)  # (frames, 180)

    # ✅ CMVN on full feature
    feats = (feats - np.mean(feats, axis=0)) / (np.std(feats, axis=0) + 1e-8)

    # Transpose → (180, frames)
    feats = feats.T

    # Add channel → (1, 180, frames)
    feats = feats[np.newaxis, :, :]

    return feats
```

## Reading Label

```python
def read_protocol(proto_path):
    """
    Returns dict: { filename : label }
    label → 0 = bonafide, 1 = spoof
    """
    label_map = {}
    with open(proto_path, "r") as f:
        for line in f:
            parts = line.strip().split()
            if len(parts) < 5:
                continue
            filename = parts[1]   # e.g. LA_T_1000137
            label    = 0 if parts[4] == "bonafide" else 1
            label_map[filename] = label
    return label_map

train_labels = read_protocol(train_proto)
dev_labels   = read_protocol(dev_proto)
eval_labels  = read_protocol(eval_proto)
```

## Dataset Class

```python
import torch
from torch.utils.data import Dataset, DataLoader

class ASVSpoofDataset(Dataset):
    def __init__(self, audio_folder, label_map, max_frames=300):
        self.max_frames = max_frames
        self.samples = []
        self.labels  = []

        for filename, label in label_map.items():
            path = os.path.join(audio_folder, filename + ".flac")
            if os.path.exists(path):
                self.samples.append(path)
                self.labels.append(label)

        print(f"Found {len(self.samples)} files | "
              f"Bonafide: {self.labels.count(0)} | "
              f"Spoof: {self.labels.count(1)}")

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        feat  = extract_lfcc(self.samples[idx])   # (1, 180, frames)
        feat  = self._fix_frames(feat)             # (1, 180, max_frames)
        label = self.labels[idx]
        return torch.tensor(feat, dtype=torch.float32), torch.tensor(label, dtype=torch.long)

    def _fix_frames(self, feat):
        T = feat.shape[2]
        if T >= self.max_frames:
            return feat[:, :, :self.max_frames]
        pad = np.zeros((1, feat.shape[1], self.max_frames - T), dtype=np.float32)
        return np.concatenate([feat, pad], axis=2)
```

## train + dev on ram cache -  eval on load


```python
from torch.utils.data import Dataset, DataLoader
# ── Cached Dataset (Train + Dev) ──────────────────────
class CachedDataset(Dataset):
    def __init__(self, audio_folder, label_map, max_frames=300):
        self.max_frames = max_frames
        self.labels = []
        self.feats  = []

        print("Pre-extracting LFCC features...")
        for filename, label in tqdm(label_map.items()):
            path = os.path.join(audio_folder, filename + ".flac")
            if not os.path.exists(path):
                continue
            try:
                feat = extract_lfcc(path)
                feat = self._fix_frames(feat)
                self.feats.append(feat)
                self.labels.append(label)
            except Exception as e:
                print(f"Error: {path} → {e}")

        print(f"Loaded {len(self.feats)} samples | "
              f"Bonafide: {self.labels.count(0)} | "
              f"Spoof: {self.labels.count(1)}")

    def __len__(self):
        return len(self.feats)

    def __getitem__(self, idx):
        return torch.tensor(self.feats[idx], dtype=torch.float32), \
               torch.tensor(self.labels[idx], dtype=torch.long)

    def _fix_frames(self, feat):
        T = feat.shape[2]
        if T >= self.max_frames:
            return feat[:, :, :self.max_frames]
        pad = np.zeros((1, feat.shape[1], self.max_frames - T), dtype=np.float32)
        return np.concatenate([feat, pad], axis=2)


```

## Cnn Model(Resnet)

```python
import torch.nn as nn


class ResBlock(nn.Module):
    def __init__(self, channels):
        super().__init__()
        self.block = nn.Sequential(
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.BatchNorm2d(channels),
            nn.ReLU(),
            nn.Conv2d(channels, channels, 3, padding=1),
            nn.BatchNorm2d(channels)
        )
        self.relu = nn.ReLU()

    def forward(self, x):
        return self.relu(x + self.block(x))  # skip connection


class SpoofCNN(nn.Module):
    def __init__(self):
        super().__init__()
        self.conv = nn.Sequential(
            # Block 1
            nn.Conv2d(1, 32, 3, padding=1), nn.BatchNorm2d(32), nn.ReLU(), nn.MaxPool2d(2),
            # Block 2
            nn.Conv2d(32, 64, 3, padding=1), nn.BatchNorm2d(64), nn.ReLU(), nn.MaxPool2d(2),
            # Block 3
            nn.Conv2d(64, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU(), nn.MaxPool2d(2),
            # Block 4 ← new
            nn.Conv2d(128, 128, 3, padding=1), nn.BatchNorm2d(128), nn.ReLU()
        )
        self.res = ResBlock(128)          # updated channels
        self.fc  = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(128, 256),          # wider
            nn.ReLU(),
            nn.Dropout(0.5),
            nn.Linear(256, 2)
        )

    def forward(self, x):
        return self.fc(self.res(self.conv(x)))
```

## Training and eval starts

```python
import torch.nn.functional as F
from sklearn.metrics import accuracy_score

# ── Loss + Optimizer + Scheduler ──────────────────────
weights   = torch.tensor([1.0, 0.5]).to(DEVICE)
criterion = nn.CrossEntropyLoss(weight=weights, label_smoothing=0.1)
optimizer = torch.optim.Adam(model.parameters(), lr=3e-4, weight_decay=1e-4)
scheduler = torch.optim.lr_scheduler.CosineAnnealingLR(optimizer, T_max=50)

# ── SpecAugment ───────────────────────────────────────
def spec_augment(feat, freq_mask=25, time_mask=50):
    feat = feat.clone()
    B, C, F, T = feat.shape
    for i in range(B):
        f  = np.random.randint(0, freq_mask)
        f0 = np.random.randint(0, F - f)
        feat[i, :, f0:f0+f, :] = 0
        t  = np.random.randint(0, time_mask)
        t0 = np.random.randint(0, T - t)
        feat[i, :, :, t0:t0+t] = 0
    return feat

# ── Evaluate Function ─────────────────────────────────
def evaluate(loader):
    model.eval()
    all_preds, all_labels = [], []
    with torch.no_grad():
        for feats, labels in loader:
            feats = feats.to(DEVICE)
            preds = model(feats).argmax(dim=1).cpu().numpy()
            all_preds.extend(preds)
            all_labels.extend(labels.numpy())
    return accuracy_score(all_labels, all_preds)

# ── Training Loop ─────────────────────────────────────
EPOCHS           = 50
best_dev_acc     = 0.0
patience         = 7
patience_counter = 0

for epoch in range(1, EPOCHS + 1):
    model.train()
    total_loss = 0

    for feats, labels in tqdm(train_loader, desc=f"Epoch {epoch}/{EPOCHS}"):
        feats, labels = feats.to(DEVICE), labels.to(DEVICE)
        feats = spec_augment(feats)

        optimizer.zero_grad()
        loss = criterion(model(feats), labels)
        loss.backward()
        optimizer.step()
        total_loss += loss.item()

    scheduler.step()
    dev_acc  = evaluate(dev_loader)
    avg_loss = total_loss / len(train_loader)
    print(f"Epoch {epoch:02d} | Loss: {avg_loss:.4f} | Dev Acc: {dev_acc*100:.2f}%")

    # ── Early Stopping ─────────────────────────────
    if dev_acc > best_dev_acc:
        best_dev_acc     = dev_acc
        patience_counter = 0
        torch.save(model.state_dict(), "best_model.pth")
        print(f"           ✅ Best model saved! Dev Acc: {dev_acc*100:.2f}%")
    else:
        patience_counter += 1
        print(f"           ⏳ No improvement. Patience: {patience_counter}/{patience}")
        if patience_counter >= patience:
            print(f"\n🛑 Early stopping at epoch {epoch}")
            break
```

```python
# ── Final Evaluation with Best Threshold ──────────────
from sklearn.metrics import confusion_matrix

THRESHOLD = 0.2

y_pred_final = (y_scores >= THRESHOLD).astype(int)
acc  = np.mean(y_true == y_pred_final)
cm   = confusion_matrix(y_true, y_pred_final)
tn, fp, fn, tp = cm.ravel()
sens = tp / (tp + fn)
spec = tn / (tn + fp)

# EER
from sklearn.metrics import roc_curve
fpr, tpr, _ = roc_curve(y_true, y_scores)
fnr = 1 - tpr
idx = np.nanargmin(np.absolute((fnr - fpr)))
eer = fpr[idx]

print("=" * 50)
print("FINAL EVALUATION RESULTS")
print("=" * 50)
print(f"Threshold   : {THRESHOLD}")
print(f"Accuracy    : {acc*100:.2f}%")
print(f"EER         : {eer*100:.2f}%")
print(f"Sensitivity : {sens*100:.2f}%")
print(f"Specificity : {spec*100:.2f}%")
print(f"\nTP: {tp}  TN: {tn}")
print(f"FP: {fp}  FN: {fn}")
print("=" * 50)
```
